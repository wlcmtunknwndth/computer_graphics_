### Важная информация по обозначениям

В печатной математической литературе не принято над векторами ставить стрелки, так как они часто мешают индексам и акцентам.  
Под акцентами понимаются дополнительные значки, проставляемые над буквой.  
Сравните: ⃗ 𝑣′ и 𝐯′ или ⃛⃗ 𝑣 и ⃛ 𝐯.

Вообще, можно выделить четыре разных начертания шрифта в формулах:

- `𝑣, 𝑝` — курсивное начертание, которым обозначают обычно **скалярные величины**;
- `v, p` — прямое начертание, которым обозначают **операторы, именные функции** (например, функции `sin 𝑥`, `tg 𝑥`, `ch x`, `ln 𝑥`);
- `𝒗, 𝒑` — полужирное курсивное начертание (**обычно не используется** в формулах);
- `𝐯, 𝐩` — полужирное прямое начертание, которым обычно обозначают **векторные величины**.

## 2.2. Задание №0

2. С помощью формулы 𝐩(𝑡) = 𝐩₁(1 − 𝑡) + 𝐩₂𝑡 отобразить точку, расположенную в середине отрезка. Векторы 𝐩₁ и 𝐩₂ — радиус-векторы точек P1 и P2 соответственно.
3. Создать ползунок, с помощью которого можно будет перемещать произвольную точку отрезка от начальной точки **P1** до конечной **P2**.

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.widgets import Slider
from ipywidgets import interact
```

```python
# 1. Задаём точки P1 и P2
P1 = np.array([1, 2])
P2 = np.array([5, 4])

def plot_point_on_segment(t=0.5, reverse=False):
    plt.figure(figsize=(6, 6))
    # Рисуем отрезок и точки
    plt.plot([P1[0], P2[0]], [P1[1], P2[1]], 'k--', label='Отрезок P1P2')
    plt.plot(P1[0], P1[1], 'ro', label='P1')
    plt.plot(P2[0], P2[1], 'bo', label='P2')
    plt.text(P1[0], P1[1], '  P1', color='r', fontsize=12)
    plt.text(P2[0], P2[1], '  P2', color='b', fontsize=12)
    # 2. Точка на отрезке
    if reverse:
        point = (1 - t) * P2 + t * P1  # от P2 к P1
    else:
        point = (1 - t) * P1 + t * P2  # от P1 к P2
    plt.plot(point[0], point[1], 'go', label='Текущая точка')
    plt.text(point[0], point[1], '  Q', color='g', fontsize=12)
    plt.legend()
    plt.xlim(min(P1[0], P2[0]) - 1, max(P1[0], P2[0]) + 1)
    plt.ylim(min(P1[1], P2[1]) - 1, max(P1[1], P2[1]) + 1)
    plt.gca().set_aspect('equal')
    plt.grid(True)
    plt.show()

# 3. Интерактивный ползунок для перемещения точки
interact(plot_point_on_segment, t=(0.0, 1.0, 0.01), reverse=False)
```

```python
# 1. Параметрическое уравнение прямой: p(t) = p0 + v * t
def line_parametric(p0, v, t):
    return p0[0] + v[0] * t, p0[1] + v[1] * t

# 2. 100 значений t на [-10, 10]
t_vals = np.linspace(-10, 10, 100)

# Начальная точка
p0 = np.array([0, 0])

def plot_line(angle_deg=30):
    angle_rad = np.deg2rad(angle_deg)
    v = np.array([np.cos(angle_rad), np.sin(angle_rad)])
    x, y = line_parametric(p0, v, t_vals)
    
    plt.figure(figsize=(6,6))
    plt.plot(x, y, label="Прямая (plot)")
    
    # 6. axline для отображения прямой (если доступно)
    ax = plt.gca()
    if hasattr(ax, "axline"):
        ax.axline(p0, slope=np.tan(angle_rad), color="red", linestyle="--", label="Прямая (axline)")
    
    plt.gca().set_aspect('equal')
    plt.xlim(-15, 15)
    plt.ylim(-15, 15)
    plt.legend()
    plt.grid(True)
    plt.title(f"Угол наклона: {angle_deg}°")
    plt.show()

# 4. Интерактивный слайдер для угла наклона
interact(plot_line, angle_deg=(0, 180, 1))
```

## 2.4. Задание №2

1. Нарисуйте без заливки цветом окружность с центром в точке **O**, радиуса 𝑅 = 1 с помощью примитива `Circle` из модуля `patches`.
2. Отобразите радиус окружности в виде стрелки или в виде отрезка.
3. Создайте ползунок, который будет вращать радиус окружности по часовой стрелке с шагом в 1°.
4. Используйте параметрическое уравнение окружности  
   𝐩(𝑡) = [𝑅 cos 𝑡, 𝑅 sin 𝑡] ⇔ {𝑥(𝑡) = 𝑅 cos 𝑡, 𝑦(𝑡) = 𝑅 sin 𝑡}  
   для того, чтобы нарисовать её с помощью функции `plot`.
5. Используйте трансляцию с помощью вектора 𝐯 = (𝑥₀, 𝑦₀)ᵀ, чтобы переместить центр окружности в точку (1, 1) по формуле:  
   𝐩′(𝑡) = 𝐩(𝑡) + 𝐯,  
   где 𝐩′ — радиус-вектор точек перемещённой окружности.

```python
from matplotlib.patches import Circle
from matplotlib.widgets import Slider

# 1. Параметры окружности
R = 1
center = np.array([0, 0])
v = np.array([1, 1])  # вектор сдвига

def plot_circles(angle_deg=0):
    angle_rad = np.deg2rad(angle_deg)
    # Окружность (исходная)
    fig, ax = plt.subplots(figsize=(6,6))
    circle = Circle(center, R, fill=False, color='blue', linewidth=2, label='Окружность (Circle)')
    ax.add_patch(circle)
    # Радиус (исходная)
    radius_end = center + R * np.array([np.cos(angle_rad), np.sin(angle_rad)])
    ax.plot([center[0], radius_end[0]], [center[1], radius_end[1]], 'r-', linewidth=2, label='Радиус')
    # Окружность (plot)
    t = np.linspace(0, 2 * np.pi, 200)
    x = R * np.cos(t)
    y = R * np.sin(t)
    ax.plot(x, y, 'g--', label='Окружность (plot)')
    # Окружность сдвинутая
    x_shifted = x + v[0]
    y_shifted = y + v[1]
    ax.plot(x_shifted, y_shifted, 'm-', label='Окружность (центр (1,1))')
    # Радиус для сдвинутой окружности
    center_shifted = center + v
    radius_end_shifted = center_shifted + R * np.array([np.cos(angle_rad), np.sin(angle_rad)])
    ax.plot([center_shifted[0], radius_end_shifted[0]], [center_shifted[1], radius_end_shifted[1]], 
            color='orange', linewidth=2, label='Радиус (центр (1,1))')
    ax.set_aspect('equal')
    ax.set_xlim(-2, 3)
    ax.set_ylim(-2, 3)
    ax.legend()
    ax.grid(True)
    plt.title(f"Угол: {angle_deg}°")
    plt.show()

interact(plot_circles, angle_deg=(0, 360, 1))
```

## 2.5. Задание №3

1. Нарисуйте прямоугольник (не квадрат) с помощью примитива `Polygon` из модуля `patches`. Нарисуйте его так, чтобы центр пересечения диагоналей лежал в начале координат.

2. Умножьте все радиус-векторы вершин прямоугольника на матрицу [( 1 -1 ) ( 1 1 )] и отобразите, во что преобразовался после этого исходный прямоугольник.

3. Вычислите ориентированную площадь исходного прямоугольника. В каком порядке нужно перечислить вершины прямоугольника, чтобы ориентированная площадь имела положительный знак?

4. Вновь умножьте радиус-векторы вершин на матрицу [( -1 1 ) ( 1 1 )] и вновь вычислите ориентированную площадь. Объясните изменение её знака и абсолютной величины.

```python
from matplotlib.patches import Polygon

# Исходные параметры прямоугольника
w, h = 4, 2
rect_vertices = np.array([
    [-w/2, -h/2],
    [ w/2, -h/2],
    [ w/2,  h/2],
    [-w/2,  h/2]
])

# Функция для вычисления ориентированной площади
def oriented_area(pts):
    x = pts[:, 0]
    y = pts[:, 1]
    return 0.5 * np.sum(x * np.roll(y, -1) - y * np.roll(x, -1))

# Возможные матрицы преобразования
matrices = {
    "Нет (исходный)": np.eye(2),
    "A = [[1, -1], [1, 1]]": np.array([[1, -1], [1, 1]]),
    "B = [[-1, 1], [1, 1]]": np.array([[-1, 1], [1, 1]])
}

def plot_transformed_rectangle(matrix_name="Нет (исходный)"):
    A = matrices[matrix_name]
    transformed_vertices = rect_vertices @ A.T
    
    fig, ax = plt.subplots(figsize=(7,7))
    
    # Исходный прямоугольник
    rect = Polygon(rect_vertices, closed=True, fill=False, edgecolor='blue', linewidth=2, label='Исходный прямоугольник')
    ax.add_patch(rect)
    ax.scatter(0, 0, color='black', label='Центр (0,0)')
    for i, (x, y) in enumerate(rect_vertices):
        ax.text(x, y, f'  P{i+1}', color='blue', fontsize=12)
    
    # Преобразованный прямоугольник
    if not np.allclose(A, np.eye(2)):
        rect_trans = Polygon(transformed_vertices, closed=True, fill=False, edgecolor='red', linestyle='--', linewidth=2, label=f'После матрицы {matrix_name}')
        ax.add_patch(rect_trans)
        for i, (x, y) in enumerate(transformed_vertices):
            ax.text(x, y, f'  Q{i+1}', color='red', fontsize=12)
    
    # Площади
    area_orig = oriented_area(rect_vertices)
    area_trans = oriented_area(transformed_vertices)
    plt.title(f"Преобразования прямоугольника\n"
              f"Площадь исходного: {area_orig:.2f}, после преобразования: {area_trans:.2f}")
    
    ax.set_aspect('equal')
    ax.legend()
    ax.set_xlim(-5, 5)
    ax.set_ylim(-5, 5)
    plt.grid(True)
    plt.show()

interact(plot_transformed_rectangle, matrix_name=list(matrices.keys()))
```

   𝐩(𝑡) = [𝑎 ch 𝑡, 𝑏 sh 𝑡] ⇔ {𝑥(𝑡) = 𝑎 ch 𝑡, 𝑦(𝑡) = 𝑏 sh 𝑡}.

4. Выясните геометрический смысл параметров 𝑎, 𝑏.

5. Создайте два ползунка, которыми можно менять значения параметров 𝑎 и 𝑏, и изучите, как при этом меняется вид кривой.

```python
from matplotlib.widgets import Slider

def plot_ellipse_hyperbola(a=2.0, b=1.0):
    t = np.linspace(0, 2 * np.pi, 100)
    x = a * np.cos(t)
    y = b * np.sin(t)
    
    fig, ax = plt.subplots(figsize=(7,7))
    # Эллипс
    ax.plot(x, y, label='Эллипс')
    # Точки эллипса
    ax.plot(x, y, marker='o', markersize=5, markevery=10, linestyle='', color='red', label='Точки эллипса')
    n_points = len(t)
    ax.set_title(f'1. Эллипс, количество значений параметра t: {n_points}')
    # Трансляция эллипса
    x_shifted = x + 1
    y_shifted = y + 1
    ax.plot(x_shifted, y_shifted, 'g--', label='Эллипс (центр (1,1))')
    # Гипербола (одна ветвь)
    t_hyper = np.linspace(-2, 2, 200)
    x_hyper = a * np.cosh(t_hyper)
    y_hyper = b * np.sinh(t_hyper)
    ax.plot(x_hyper, y_hyper, 'm-', label='Гипербола (одна ветвь)')
    ax.set_aspect('equal')
    ax.set_xlim(-5, 5)
    ax.set_ylim(-5, 5)
    ax.legend()
    ax.grid(True)
    plt.show()

interact(plot_ellipse_hyperbola, a=(0.5, 5.0, 0.1), b=(0.5, 5.0, 0.1))
```

## 2.7. Задание №5

1. Используя код из задания №1, нарисуйте две прямые и вычислите угол между ними, то есть угол между направляющими векторами 𝐯₁ и 𝐯₂ первой и второй прямых соответственно.

2. Для вычисления угла используйте функцию `arctan2` из библиотеки NumPy или `atan2` из библиотеки math.

```python
import numpy as np
import matplotlib.pyplot as plt

def plot_two_lines(v1_x=2.0, v1_y=1.0, v2_x=1.0, v2_y=2.0):
    p0_1 = np.array([0, 0])
    v1 = np.array([v1_x, v1_y])
    p0_2 = np.array([0, 0])
    v2 = np.array([v2_x, v2_y])
    t = np.linspace(-10, 10, 100)
    line1_x = p0_1[0] + v1[0] * t
    line1_y = p0_1[1] + v1[1] * t
    line2_x = p0_2[0] + v2[0] * t
    line2_y = p0_2[1] + v2[1] * t

    fig, ax = plt.subplots(figsize=(7,7))
    ax.plot(line1_x, line1_y, label='Прямая 1')
    ax.plot(line2_x, line2_y, label='Прямая 2')
    ax.quiver(*p0_1, *v1, angles='xy', scale_units='xy', scale=1, color='blue', label='v1')
    ax.quiver(*p0_2, *v2, angles='xy', scale_units='xy', scale=1, color='red', label='v2')
    ax.set_aspect('equal')
    ax.set_xlim(-10, 10)
    ax.set_ylim(-10, 10)
    ax.legend()
    # 2. Угол между векторами
    dot = np.dot(v1, v2)
    det = v1[0]*v2[1] - v1[1]*v2[0]
    angle_rad = np.arctan2(abs(det), dot)
    angle_deg = np.degrees(angle_rad)
    plt.title(f"Две прямые и их направляющие векторы\nУгол между прямыми: {angle_deg:.2f}°")
    plt.grid(True)
    plt.show()

interact(plot_two_lines, 
         v1_x=(-5.0, 5.0, 0.1), 
         v1_y=(-5.0, 5.0, 0.1), 
         v2_x=(-5.0, 5.0, 0.1), 
         v2_y=(-5.0, 5.0, 0.1))
```
